# 01 - Data Exploration

Explore the JobPulse AI dataset: structure, missing values, duplicates, and key distributions.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
pd.set_option('display.max_columns', None)

from src.data_loader import load_raw_data, detect_and_rename_columns
from src.data_cleaning import clean_data
from src.skill_extractor import extract_skills
from src.feature_engineering import engineer_features
```

```python
# Load raw data
df = load_raw_data()
df = detect_and_rename_columns(df)
df.head()
```

## Dataset Overview

```python
print(f"Shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
df.info()
```

## Missing Values

```python
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
missing.plot(kind='barh', figsize=(10, 6), title='Missing values by column')
plt.tight_layout()
plt.show()
print(missing)
```

## Duplicate Analysis

```python
exact_dupes = df.duplicated(keep='first').sum()
print(f"Exact duplicates: {exact_dupes}")
print(f"Duplicate %: {exact_dupes / len(df) * 100:.2f}%")
```

## Job Role Distribution

```python
cleaned = clean_data(df)
cleaned['standardized_job_title'].value_counts().plot(kind='bar', figsize=(10, 5), title='Jobs by canonical role')
plt.tight_layout()
plt.show()
```

## Location Distribution

```python
cleaned['city'].value_counts().head(10).plot(kind='bar', figsize=(10, 5), title='Top 10 cities')
plt.tight_layout()
plt.show()
```

## Company Distribution

```python
cleaned['company'].value_counts().head(15).plot(kind='bar', figsize=(10, 6), title='Top 15 companies')
plt.tight_layout()
plt.show()
```
